```python
%sql
WITH bronze as(
    SELECT COUNT (*) as total_bronze
    FROM kipu_pay.bronze.refunds_chargebacks
), 
landing as (
    SELECT COUNT(*) as total_landing
    FROM read_files (
        '/Volumes/kipu_pay/landing/raw/refunds_chargebacks.csv',
        format => 'csv',
        header => true,
        inferSchema => false)
) 
SELECT bronze.total_bronze, landing.total_landing,
CASE WHEN bronze.total_bronze = landing.total_landing THEN 'OK' ELSE 'NOT OK' END as status
FROM bronze
CROSS JOIN landing
```
