```python
 %sql

WITH bronze AS (
    SELECT COUNT(*) AS total_bronze
    FROM kipu_pay.bronze.app_events
),

landing AS (
    SELECT COUNT(*) AS total_landing
     FROM read_files(
        '/Volumes/kipu_pay/landing/raw/app_events_2026-11-09_22.jsonl', 
        format => 'json',
        inferSchema => true 
    )
)

SELECT
    landing.total_landing,
    bronze.total_bronze,
    CASE
        WHEN landing.total_landing = bronze.total_bronze THEN 'OK'
        ELSE 'MISMATCH'
    END AS resultado
FROM bronze
CROSS JOIN landing;

 
 
 
```
