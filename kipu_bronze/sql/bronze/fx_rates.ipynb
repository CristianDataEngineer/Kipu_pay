```python
%sql

WITH bronze AS (
    SELECT COUNT(*) AS total_bronze
    FROM kipu_pay.bronze.fx_rates
),

landing AS (
    SELECT COUNT(*) AS total_landing
    FROM read_files(
        '/Volumes/kipu_pay/landing/raw/fx_rates.csv',
        format => 'csv',
        header => true,
        inferSchema => false 
    )
)

SELECT
    landing.total_landing,
    bronze.total_bronze,
    CASE
        WHEN landing.total_landing = bronze.total_bronze THEN 'OK'
        ELSE 'MISMATCH'
    END AS resultado
FROM bronze
CROSS JOIN landing;
```
