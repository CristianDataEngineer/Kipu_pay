```python
%sql

INSERT OVERWRITE kipu_pay.bronze.users
    SELECT
        user_id,
        email,
        country,
        city,
        signup_ts,
        kyc_level,
        birth_year,
        updated_at
    FROM read_files(
    '/Volumes/kipu_pay/landing/raw/users.csv',
    format => 'csv',
    header => true,
    inferSchema => false
);
```
